# Business Rules & Recommendation Engine

## Objective

This notebook implements business logic on top of machine learning predictions.

The recommendation engine provides:

- Inventory Status
- Promotion Recommendations
- Pricing Recommendations

These outputs will later be integrated into the web application dashboard.

## Inventory Recommendation

Determines whether stock levels are sufficient
based on predicted demand.

```python
def inventory_status(
    inventory,
    predicted_sales
):
    
    if predicted_sales > inventory:
        return "⚠️ Restock Required"
    
    return "✅ Inventory Sufficient"
```

## Promotion Recommendation

Suggests promotional activities when
predicted demand is below forecasted demand.

```python
def promotion_recommendation(
    predicted_sales,
    demand_forecast
):
    
    if predicted_sales < demand_forecast:
        return "📢 Run Promotion"
    
    return "✅ No Promotion Needed"
```

## Pricing Recommendation

Compares product price with competitor pricing.

```python
def pricing_recommendation(
    price,
    competitor_price
):
    
    if price > competitor_price:
        return "💰 Price Above Competitor"
    
    elif price < competitor_price:
        return "🔥 Competitive Pricing Advantage"
    
    else:
        return "⚖️ Market Matching Price"
```

## Inventory Risk Assessment

Classifies inventory risk based on
inventory coverage.

```python
def inventory_risk(
    inventory,
    predicted_sales
):
    
    ratio = inventory / (predicted_sales + 1)
    
    if ratio < 0.8:
        return "🔴 High Risk"
    
    elif ratio < 1.2:
        return "🟠 Medium Risk"
    
    return "🟢 Low Risk"
```

```python
predicted_sales = 150

inventory = 120

demand_forecast = 170

price = 35

competitor_price = 30
```

```python
print(
    inventory_status(
        inventory,
        predicted_sales
    )
)

print(
    promotion_recommendation(
        predicted_sales,
        demand_forecast
    )
)

print(
    pricing_recommendation(
        price,
        competitor_price
    )
)

print(
    inventory_risk(
        inventory,
        predicted_sales
    )
)
```
