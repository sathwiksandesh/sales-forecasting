# Model Training & Comparison

## Objective

The objective of this notebook is to train multiple machine learning algorithms and compare their performance for retail sales prediction.

Models Evaluated:

- Linear Regression
- Random Forest
- Extra Trees
- Gradient Boosting
- XGBoost
- LightGBM
- CatBoost

Evaluation Metrics:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R² Score

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.linear_model import LinearRegression

from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor
)

from xgboost import XGBRegressor

from lightgbm import LGBMRegressor

from catboost import CatBoostRegressor

import warnings
warnings.filterwarnings("ignore")
```

```python
X_train = joblib.load("X_train.pkl")
X_test = joblib.load("X_test.pkl")

y_train = joblib.load("y_train.pkl")
y_test = joblib.load("y_test.pkl")
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train, y_train)

pred_lr = lr.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred_lr))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_lr)))
print("R2:", r2_score(y_test, pred_lr))
```

# Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

pred_rf = rf.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred_rf))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_rf)))
print("R2:", r2_score(y_test, pred_rf))
```

# Extra Trees

```python
from sklearn.ensemble import ExtraTreesRegressor

et = ExtraTreesRegressor(
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

et.fit(X_train, y_train)

pred_et = et.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred_et))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_et)))
print("R2:", r2_score(y_test, pred_et))
```

# XGBoost

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    objective="reg:squarederror"
)

xgb.fit(X_train, y_train)

pred_xgb = xgb.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred_xgb))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_xgb)))
print("R2:", r2_score(y_test, pred_xgb))
```

# LightGBM

```python
from lightgbm import LGBMRegressor

lgbm = LGBMRegressor(
    random_state=42
)

lgbm.fit(X_train, y_train)

pred_lgbm = lgbm.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred_lgbm))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_lgbm)))
print("R2:", r2_score(y_test, pred_lgbm))
```

# Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(
    random_state=42
)

gb.fit(
    X_train,
    y_train
)

pred_gb = gb.predict(X_test)

print("Gradient Boosting")

print(
    "MAE:",
    mean_absolute_error(y_test, pred_gb)
)

print(
    "RMSE:",
    np.sqrt(
        mean_squared_error(
            y_test,
            pred_gb
        )
    )
)

print(
    "R2:",
    r2_score(
        y_test,
        pred_gb
    )
)
```

```python
results = []

results.append([
    "Linear Regression",
    mean_absolute_error(y_test, pred_lr),
    np.sqrt(mean_squared_error(y_test, pred_lr)),
    r2_score(y_test, pred_lr)
])

results.append([
    "Random Forest",
    mean_absolute_error(y_test, pred_rf),
    np.sqrt(mean_squared_error(y_test, pred_rf)),
    r2_score(y_test, pred_rf)
])

results.append([
    "Extra Trees",
    mean_absolute_error(y_test, pred_et),
    np.sqrt(mean_squared_error(y_test, pred_et)),
    r2_score(y_test, pred_et)
])

results.append([
    "Gradient Boosting",
    mean_absolute_error(y_test, pred_gb),
    np.sqrt(mean_squared_error(y_test, pred_gb)),
    r2_score(y_test, pred_gb)
])

results.append([
    "XGBoost",
    mean_absolute_error(y_test, pred_xgb),
    np.sqrt(mean_squared_error(y_test, pred_xgb)),
    r2_score(y_test, pred_xgb)
])

results.append([
    "LightGBM",
    mean_absolute_error(y_test, pred_lgbm),
    np.sqrt(mean_squared_error(y_test, pred_lgbm)),
    r2_score(y_test, pred_lgbm)
])
```
